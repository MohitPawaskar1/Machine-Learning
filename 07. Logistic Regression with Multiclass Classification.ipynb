```python
# Importing the Required Models

import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
import seaborn as sns
```

```python
# Reading the Dataset

digits = load_digits()
```

```python
dir(digits)
```

```python
digits.data[0]
```

```python
plt.gray()
for i in range(5):
    plt.matshow(digits.images[10])
```

```python
digits.target[0:5]
```

```python
# Spliting the Data into Training and Testing

X_train, X_test, y_trian, y_test = train_test_split(digits.data, digits.target, test_size=0.2)
```

```python
len(X_test)
```

```python
# Building the Model

model = LogisticRegression(max_iter=10000)
model.fit(X_train, y_trian)
```

```python
# Seeing the Score of the Model

model.score(X_test, y_test)
```

```python
plt.matshow(digits.images[69])
```

```python
digits.target[69]
```

```python
# Predicting the Values

model.predict([digits.data[69]])
```

```python
# Plot digits 90 to 100

fig, axes = plt.subplots(1, 11, figsize=(12, 3))
for i, ax in enumerate(axes):
    ax.matshow(digits.images[90 + i], cmap='gray')
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Predicting the Values

model.predict(digits.data[90:101])
```

```python
# Declaring the y pred to plot the confusion matrix

y_predicted = model.predict(X_test)
```

```python
# Creating the Confusion Matrix 

cm = confusion_matrix(y_test, y_predicted)
cm
```

```python
# Ploting the Confusion Matrix


plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True)
plt.xlabel('predicted')
plt.ylabel('Truth')
```
